```python
def import_data_proj(PATH):

    df = pd.read_excel(PATH)

    # Excluindo header e arrumando cabeçalho
    df = df.iloc[5:].reset_index(drop=True)
    df.columns = df.iloc[0]
    df = df.iloc[1:].reset_index(drop=True)

    cols = df.columns[4:69]
    df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')
    df['ANO'] = df['ANO'].astype(int)

    return df

def import_data_fin(PATH):

    sheets = pd.read_excel(PATH, sheet_name=None)

    dfs = []

    for i, (nome, df) in enumerate(sheets.items()):
        
        # Transpondo df
        if nome == '22C40_01' or nome == '22C40_01 2' or nome == '22C40_02' or nome == '22C40_02 2':
            if i == 0:
                df = df.T.iloc[:, [3, 4, 20, 29, 40]].reset_index(drop=True)
            else:
                df = df.T.iloc[:, [4, 5, 21, 30, 41]].reset_index(drop=True)
        else:
            if i == 0:
                df = df.T.iloc[:, [3, 4, 20, 29, 44]].reset_index(drop=True)
            else:
                df = df.T.iloc[:, [4, 5, 21, 30, 45]].reset_index(drop=True)

        # Ajustando cabeçalho
        df = df.iloc[1:].reset_index(drop=True)
        df.columns = ['Mes', 'Arrecadacao_Bruta', 'Arrecadacao_Liquida', 'Total_Beneficios', 'Resultado_Primario']
        df = df.rename(columns={float('nan'): 'Mes'})

        # Convertendo pra numerico
        cols = df.columns[1:5]
        df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')

        dfs.append(df)

    # Concatenando as abas
    df_final = pd.concat(dfs, ignore_index=True)

    return df_final
```

**evolução de idosos ao longo dos anos**

```python
anos = sorted(PROJECAO_2024["ANO"].unique())

indicadores = []

for ano in anos:
    dados = PROJECAO_2024[PROJECAO_2024["ANO"] == ano]

    idosos = dados["65+_T"].sum()
    total = dados["POP_T"].sum()

    indicadores.append((idosos / total) * 100)

plt.figure(figsize=(10, 5))

plt.plot(anos, indicadores, marker="o")

plt.title("Evolução da população com 65 anos ou mais")
plt.xlabel("Ano")
plt.ylabel("População 65+ (%)")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

**idosos por 100 pessoas em idade ativa**

```python
anos = sorted(PROJECAO_2024["ANO"].unique())

dependencia = []

for ano in anos:
    dados = PROJECAO_2024[PROJECAO_2024["ANO"] == ano]

    nao_ativos = dados["65+_T"].sum() + dados["0-14_T"].sum()
    populacao_ativa = dados["15-64_T"].sum()

    dependencia.append((nao_ativos / populacao_ativa) * 100)

plt.figure(figsize=(10, 5))

plt.plot(anos, dependencia, marker="o")

plt.title("Evolução da razão de dependência")
plt.xlabel("Ano")
plt.ylabel("Pessoas não ativas por 100 pessoas em idade ativa")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Abreviações dos meses
meses = {
    "Janeiro": "jan",
    "Fevereiro": "fev",
    "Março": "mar",
    "Abril": "abr",
    "Maio": "mai",
    "Junho": "jun",
    "Julho": "jul",
    "Agosto": "ago",
    "Setembro": "set",
    "Outubro": "out",
    "Novembro": "nov",
    "Dezembro": "dez"
}

dfs = []

for ano in range(2020, 2026):

    df = globals()[f"FINANCAS_INSS_{ano}"].copy()

    # Remove a linha de total anual
    df = df[df["Mes"].str.strip().str.casefold() != "total"].copy()

    # Adiciona o ano
    df["ANO"] = ano

    # Cria o número do mês
    df["MES_NUM"] = df["Mes"].map({
        nome: numero
        for numero, nome in enumerate(meses.keys(), start=1)
    })

    # Renomeia os meses para jan/2020, fev/2020...
    df["Mes"] = df["Mes"].map(meses) + "/" + str(ano)

    dfs.append(df)

# Concatena todos os DataFrames
FINANCAS_INSS = pd.concat(dfs, ignore_index=True)

# Ordena cronologicamente
FINANCAS_INSS = FINANCAS_INSS.sort_values(
    ["ANO", "MES_NUM"]
).reset_index(drop=True)

FINANCAS_INSS.to_csv(
    "../data/transformed_data/financas_inss.csv",
    index=False,
    encoding="utf-8-sig"
)
```
